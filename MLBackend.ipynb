Setup

```python
from pathlib import Path
import pandas as pd
import kagglehub
```

```python


dataset_dir = Path(
    kagglehub.dataset_download(
        "ealtman2019/ibm-transactions-for-anti-money-laundering-aml"
    )
)

csv_paths = sorted(dataset_dir.rglob("*.csv"))

if not csv_paths:
    raise FileNotFoundError(f"No CSV files found in {dataset_dir}")

print(f"Downloaded {len(csv_paths)} CSV files:")
for path in csv_paths:
    print(path.name)



```

```python
# load HI-SMALL (accounts and transactions)

# Grab the two HI-Small paths from csv_paths
trans_path = next(p for p in csv_paths if p.name == "HI-Small_Trans.csv")

# Transactions
trans_dtypes = {
    "From Bank": "int32",
    "To Bank": "int32",
    "Account": "string",
    "Account.1": "string",
    "Amount Received": "float64",
    "Amount Paid": "float64",
    "Receiving Currency": "category",
    "Payment Currency": "category",
    "Payment Format": "category",
    "Is Laundering": "int8",
}
trans_df = pd.read_csv(trans_path, dtype=trans_dtypes, parse_dates=["Timestamp"])


print("transactions:", trans_df.shape)
display(trans_df[trans_df["Is Laundering"] == 1].sample(10))
```
